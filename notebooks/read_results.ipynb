## 3. Résultats de la Simulation de Maintenance

```python
# Lire le résumé de la simulation
with open(os.path.join(results_dir, 'simulator_results.txt'), 'r') as f:
    print(f.read())

# Charger les résultats de simulation
sim_results = pd.read_csv(os.path.join(results_dir, 'simulation_results.csv'))
print("\nAperçu des résultats de simulation:")
sim_results.head(10)
```

```python
# Statistiques des anomalies
total_samples = len(sim_results)
anomalies = (sim_results['anomaly'] == -1).sum()
normal = (sim_results['anomaly'] == 1).sum()

print(f"Total échantillons: {total_samples}")
print(f"Anomalies détectées: {anomalies}")
print(f"Échantillons normaux: {normal}")
print(f"Pourcentage d'anomalies: {anomalies/total_samples*100:.1f}%")
```

```python
# Afficher le graphique de simulation
img = mpimg.imread(os.path.join(results_dir, 'simulation_plot.png'))
plt.figure(figsize=(15, 8))
plt.imshow(img)
plt.axis('off')
plt.title('Simulation de Maintenance Prédictive - Détection d\'Anomalies')
plt.show()
```
